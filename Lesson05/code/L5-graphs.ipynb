```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

mpg = sns.load_dataset('mpg')
mpg['origin'] = mpg['origin'].map({'usa':'USA','japan':'Japan','europe':'Europe'})
mpg['model_year'] = mpg['model_year'] + 1900
```

## 5.1 What is a Histogram

```python
sns.histplot(x='mpg', data=mpg, binwidth=5)
plt.title('Distribution of MPG')
plt.savefig('../images/Fig5_1_Histogram.png')
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
sns.countplot(x='origin', data=mpg, ax=ax[0])
ax[0].set_title('Count of Cars by Origin')
sns.histplot(x='mpg', data=mpg, ax=ax[1], binwidth=5)
ax[1].set_title('Distribution of MPG')
plt.savefig('../images/Fig5_1_BarVsHistogram.png')
```

## 5.2 Creating a Histogram

```python
d = [5, 7, 9, 12, 14, 16, 18, 20, 23, 26, 29, 32, 35, 37, 38, 41, 44, 45]

sns.histplot(d, bins=5, binwidth=10)
plt.title('Time to Commute to Campus')
plt.savefig('../images/Fig5_2_CommuteHistogram.png')
```
