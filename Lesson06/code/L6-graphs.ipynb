```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Practice 6.5.1
stats = [10, 12, 13, 14, 15, 18.5, 20, 55]

plt.figure(figsize=(8, 1))
sns.boxplot(x=stats)
plt.title('Boxplot for 5-number Summary in Practice Question 1')
plt.yticks([])
plt.xticks(np.arange(8, 59, 2))
plt.xlabel('Values')

plt.savefig('../images/Fig651_Solution.png', bbox_inches='tight')
```
