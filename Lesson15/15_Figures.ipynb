## 15.1 Discrete and Contiuous Variables

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
n = 30000
B = np.random.binomial(n=12, p=0.5, size=n)
sns.barplot(np.bincount(B), color='#3274a1')
print(12*0.5, np.sqrt(12*0.5*0.5))
plt.title("30,000 samples from a Binomial Distribution")
plt.xlabel("Value")
plt.ylabel("Count")
plt.show()
```

```python
import pandas as pd
U = pd.DataFrame()
U.index = [1,2,3,4,5,6]
U['value'] = 1/6
U
sns.barplot(x=U.index, y=U['value'], color='#3274a1')
plt.title("Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.barplot(x=U.index, y=U['value'], fill=False, color='#3274a1')
sns.barplot(x=U.loc[:4].index, y=U.loc[:4]['value'], color='#3274a1')
plt.title("Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

```python
U2 = np.arange(1,6,0.0001)
sns.histplot(U2, bins=12, stat='density', color='#3274a1')
plt.title("a Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

```python
sns.histplot(U2, bins=30, stat='density', color='#3274a1')
plt.title("a Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

```python
sns.histplot(U2, bins=50, stat='density', color='#3274a1')
plt.title("a Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

```python
sns.histplot(U2, bins=120, stat='density', color='#3274a1')
plt.title("a Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

```python
sns.histplot(U2, bins=120, stat='density', color='#3274a1', fill=False)
plt.fill_betweenx([0,0.20], 1, 4, color='#3274a1', alpha=0.8)
plt.title("a Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

```python
U2 = np.random.uniform(low=1, high=6, size=n)
sns.histplot(U2, bins=12, stat='proportion', color='#3274a1')
plt.title("30,000 samples from a Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Proportion")
plt.show()
```

```python
sns.histplot(U2, bins=50, stat='proportion', color='#3274a1')
plt.title("30,000 samples from a Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Proportion")
plt.show()
```

```python
sns.histplot(U2, bins=120, stat='proportion', color='#3274a1')
plt.title("30,000 samples from a Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Proportion")
plt.show()
```

```python
sns.histplot(U2, bins=120, stat='proportion', color='#3274a1', fill=False)
sns.histplot(U2[U2 <= 4], bins=120, stat='proportion', color='#3274a1', fill=True)
plt.title("30,000 samples from a Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Proportion")
plt.show()
```
